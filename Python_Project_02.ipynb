# 1. Contact Book

```python
# Initialize contacts
contacts = {}
# menu
def show_menu():
  print("\n----Contact Book Menu:-----")
  print("1. Show Contact list")
  print("2. Add Contact")
  print("3. Search Contact")
  print("4. Edit Contact")
  print("5. Delete Contact")
  print("6. Exit")
  print("----------------------------")
  choice = input("Enter your choice: ")
  return choice

# funstions
def show_contacts():
  if contacts:
    print("\n----Contact List:-----")
    for name, contact_details in contacts.items():
      print(f"Name: {name}")
      print(f"Phone: {contact_details['phone']}")
      print(f"Email: {contact_details['email']}")
      print("----------------------------")
  else:
    print("--------------------------")
    print("(Empty)No contacts found.")


def add_contact():
  name = input("Enter name: ")
  phone = input("Enter phone number: ")
  email = input("Enter email: ")
  contacts[name] = {"phone": phone, "email": email}
  print(f"{name} has been added to the contact list.")


def search_contact():
  name = input("Enter Contact name: ")
  if name in contacts:
    print(f"\n----Contact Details for {name}:-----")
    print(f"Name: {name}")
    print(f"Phone: {contacts[name]['phone']}")
    print(f"Email: {contacts[name]['email']}")
  else:
    print(f"{name} is not in the contact list.")


def edit_contact():
  name = input("Enter Contact name you want to edit: ")
  if name in contacts:
    phone = input("Enter new phone number: ")
    email = input("Enter new email: ")
    contacts[name] = {"phone": phone, "email": email}
    print(f"{name}'s contact has been updated.")
  else:
    print(f"{name} is not in the contact list.")


def delete_contact():
  name = input("Enter Contact name: ")
  if name in contacts:
    del contacts[name]
    print(f"{name} has been deleted from the contact list.")
  else:
    print(f"{name} is not found in the contact list.")

# Main program
while True:
  choice = show_menu()
  if choice == "1":
    show_contacts()
  elif choice == "2":
    add_contact()
  elif choice == "3":
    search_contact()
  elif choice == "4":
    edit_contact()
  elif choice == "5":
    delete_contact()
  elif choice == "6":
    print("Exiting Contact Book. Goodbye!")
    break
  else:
    print("Invalid choice. Please select a valid option.")
```

# 2. Ingridient Chacker

```python
# ingridient for recipe
recipe_ingredients = {
    "Pizza": ["dough", "tomato sauce", "cheese", "pepperoni"],
    "Burger": ["bun", "beef patty", "lettuce", "tomato"],
    "Salad": ["lettuce", "tomato", "cucumber", "oil"],
    "Pasta": ["pasta", "tomato sauce", "cheese", "parmesan"],
    "Soup": ["water", "lentils", "tomato", "onion"]
}
# user input
user_input_raw = input("Enter a recipe: ")
user_ingredients_raw = input("Enter ingredients you have (seperated by comma): ")

# Process inputs
user_input_processed = user_input_raw.title() # Convert to title case for matching recipe keys
user_ingredients = {ingredient.strip() for ingredient in user_ingredients_raw.split(",") if ingredient.strip()}  # Strip whitespace and convert to set, ignore empty strings

# Get the required ingredients for the recipe
required_ingredients = set(recipe_ingredients.get(user_input_processed, []))

# compare
if required_ingredients and required_ingredients.issubset(user_ingredients):
  print(f"You can make {user_input_processed}.")
  missing_ingredients = required_ingredients - user_ingredients
  print(f"Missing ingredients: {', '.join(missing_ingredients) if missing_ingredients else 'None'}")
  extra_ingredients = user_ingredients - required_ingredients
  print(f"Extra ingredients: {', '.join(extra_ingredients) if extra_ingredients else 'None'}")
elif not required_ingredients: # Handle case where recipe is not found
    print(f"Recipe '{user_input_processed}' not found in the list.")
    print(f"You have these ingredients: {', '.join(user_ingredients) if user_ingredients else 'None'}.")
else:
  print(f"You cannot make {user_input_processed}.")
  missing_ingredients = required_ingredients - user_ingredients
  print(f"Missing ingredients: {', '.join(missing_ingredients) if missing_ingredients else 'None'}")
  extra_ingredients = user_ingredients - required_ingredients
  print(f"Extra ingredients: {', '.join(extra_ingredients) if extra_ingredients else 'None'}")
```

# 3. Note-Taking app

```python
# initialize filename
filename = "notes.txt"

# menu
def show_menu():
    print("\n---- Note-Taking Menu -----")
    print("1. Show Notes")
    print("2. Add Note")
    print("3. Search Note")
    print("4. Edit Note")
    print("5. Delete Note")
    print("6. Exit")
    print("---------------------------")


def show_notes():
    try:
        with open(filename, "r") as file:
            notes = file.read().splitlines()

        if notes:
            print("\n---- My Notes -----")
            for i, note in enumerate(notes, 1):
                print(f"{i}. {note}")
        else:
            print("\nNo Note Found!")

    except FileNotFoundError:
        print("No Note Found!")


def add_note():
    note = input("Enter a new note: ")

    if note.strip() == "":
        print("Empty note cannot be added.")
        return

    with open(filename, "a") as file:
        file.write(note + "\n")

    print("Note added successfully!")


def search_note():
    search_term = input("Enter a search term: ")

    try:
        with open(filename, "r") as file:
            notes = file.read().splitlines()

        matching_notes = [note for note in notes if search_term.lower() in note.lower()]

        if matching_notes:
            print("\n---- Matching Notes -----")
            for note in matching_notes:
                print(note)
        else:
            print("No matching notes found.")

    except FileNotFoundError:
        print("No Note Found!")


def edit_note():
    try:
        with open(filename, "r") as file:
            notes = file.read().splitlines()

        if not notes:
            print("No notes to edit.")
            return

        print("\n---- Notes to Edit -----")
        for i, note in enumerate(notes, 1):
            print(f"{i}. {note}")

        try:
            num = int(input("Enter note number to edit: "))
        except ValueError:
            print("Please enter a valid number.")
            return

        if 1 <= num <= len(notes):
            new_note = input("Enter new text: ")
            notes[num - 1] = new_note

            with open(filename, "w") as file:
                file.write("\n".join(notes) + "\n")

            print("Note updated successfully!")
        else:
            print("Invalid note number.")

    except FileNotFoundError:
        print("No Note Found!")


def delete_note():
    try:
        with open(filename, "r") as file:
            notes = file.read().splitlines()

        if not notes:
            print("No notes to delete.")
            return

        print("\n---- Notes to Delete -----")
        for i, note in enumerate(notes, 1):
            print(f"{i}. {note}")

        try:
            num = int(input("Enter note number to delete: "))
        except ValueError:
            print("Please enter a valid number.")
            return

        if 1 <= num <= len(notes):
            notes.pop(num - 1)

            with open(filename, "w") as file:
                file.write("\n".join(notes) + "\n")

            print("Note deleted successfully!")
        else:
            print("Invalid note number.")

    except FileNotFoundError:
        print("No Note Found!")


# main program
while True:
    show_menu()
    choice = input("Enter your choice: ")

    if choice == "1":
        show_notes()

    elif choice == "2":
        add_note()

    elif choice == "3":
        search_note()

    elif choice == "4":
        edit_note()

    elif choice == "5":
        delete_note()

    elif choice == "6":
        print("Exiting Note-Taking App. Goodbye!")
        break

    else:
        print("Invalid choice. Please select a valid option.")
```

# 4. Safe Calculator

```python
# define calculator's functions
def add(x, y):
  return x + y

def subtract(x, y):
  return x - y

def multiply(x, y):
  return x * y

def divide(x, y):
  try:
    return x / y
  except ZeroDivisionError:
    return "Cannot divide by zero."

def modulus(x, y):
  return x % y


def show_menu():
  print("\n----Safe Calculator Menu:-----")
  print("1. Addition")
  print("2. Subtraction")
  print("3. Multiplication")
  print("4. Division")
  print("5. Modulus")
  print("6. Exit")
  print("----------------------------")

# main program
while True:
  show_menu()
  choice = input("Enter your choice(1-6): ")

  if choice == "6":
    print("Exiting Safe Calculator. Goodbye!")
    break

  try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    if choice == "1":
      print("Result : ", add(num1, num2))
    elif choice == "2":
      print("Result : ", subtract(num1, num2))
    elif choice == "3":
      print("Result : ", multiply(num1, num2))
    elif choice == "4":
      print("Result : ", divide(num1, num2))
    elif choice == "5":
      print("Result : ", modulus(num1, num2))
    else:
      print("Invalid choice. Please select a valid option.")

  except ValueError:
    print("Invalid input. Please enter valid numbers.")
  except ZeroDivisionError as e:
    print(f"Error: {e}")
  except Exception as e:
    print(f"An exception error occurred: {e}")
  finally:
    print("Exicution completed. Thanks for using our safe calculator.")
```

# 5. Temperature Converter

```python
#step 1: define functions
def celsius_to_fahrenheit(celsius):
  fahrenheit = (celsius * 9/5) + 32
  return fahrenheit

def celsius_to_kelvin(celsius):
  return celsius + 273.15

def fahrenheit_to_celsius(fahrenheit):
  celsius = (fahrenheit - 32) * 5/9
  return celsius

def fahrenheit_to_kelvin(fahrenheit):
  return (fahrenheit -32) * 5/9 + 273.15

def kelvin_to_celsius(kelvin):
  return kelvin - 273.15

def kelvin_to_fahrenheit(kelvin):
  return (kelvin - 273.15) * 9/5 + 32

#step 2: show menu
def show_menu():
  print("\n----Temperature Converter-----")
  print("1. Celsius to Fahrenheit")
  print("2. Celsius to kelvin")
  print("3. Fahrenheit to Celsius")
  print("4. Fahrenheit to kelvin")
  print("5. kelvin to Celsius")
  print("6. kelvin to Fahrenheit")
  print("7. EXIT")

#step 3:main program
while True:
  show_menu()
  choice = input("Enter your choice(1-6): ")

  if choice == '1':
    celsius = float(input("Enter temperature in celsius: "))
    print(f"Fahrenheit: {celsius_to_fahrenheit(celsius):.2f}")
  elif choice == '2':
    celsius = float(input("Enter temperature in celsius: "))
    print(f"Kelvin: {celsius_to_kelvin(celsius):.2f}")
  elif choice == '3':
    fahrenheit = float(input("Enter temperature in fahrenheit: "))
    print(f"Celsius: {fahrenheit_to_celsius(fahrenheit):.2f}")
  elif choice == '4':
    fahrenheit = float(input("Enter temperature in fahrenheit: "))
    print(f"Kelvin: {fahrenheit_to_kelvin(fahrenheit):.2f}")
  elif choice == '5':
    kelvin = float(input("Enter temperature in kelvin: "))
    print(f"Celsius: {kelvin_to_celsius(kelvin):.2f}")
  elif choice == '6':
    kelvin = float(input("Enter temperature in kelvin: "))
    print(f"Fahrenheit: 7{kelvin_to_fahrenheit(kelvin):.2f}")
  elif choice == '7':
    print("Exiting the program.")
    break
  else:
    print("Please enter a valid input.")

```
